```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the CSV
data = pd.read_csv("20250810_funds_open-end-fund-profile.csv")

# Quick check of the column names
print(data.columns)

# Make sure the columns exist and have numeric data
x = data["Fund AuM (m)"]
y = data["Monthly Return Jul 2025"]

# Basic scatter plot (better for continuous variables)
plt.figure(figsize=(8,5))
plt.scatter(x, y)
plt.xlabel("Fund AuM (millions)")
plt.ylabel("Monthly Return (July 2025)")
plt.title("Relationship between Fund AuM and July 2025 Monthly Return")
plt.show()
```
